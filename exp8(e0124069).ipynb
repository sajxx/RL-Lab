```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

dataset = pd.read_csv('Ads_CTR_Optimisation.csv')

print(dataset.describe())

import random

N = 10000
D = 10

numbers_of_selections = [0] * D
Sum_of_rewards = [0] * D
ads_selected = []
total_reward = 0

epsilon = 0.1

mean_rewards = dataset.describe().loc['mean'].values

for n in range(0, N):
    ad = 0
    if random.uniform(0, 1) > epsilon:
        ad = random.randrange(D)
    else:
        ad = np.argmax(mean_rewards)

    ads_selected.append(ad)
    number_of_rewards = dataset.values[n, ad]
    Sum_of_rewards[ad] += number_of_rewards
    numbers_of_selections[ad] += 1
    total_reward += number_of_rewards

plt.bar(range(D), numbers_of_selections)
plt.xticks(range(D))
plt.xlabel('Advertisement')
plt.ylabel('Number of selections')
plt.title('Advertisement Selection Bar Chart')
plt.show()

        
```
